```python
input_file = "executed_output.txt"
output_file = "consec_grouped_summary.txt"

# read file
with open(input_file, "r") as f:
    lines = [line.strip() for line in f]

# remove numbering (everything before first colon)
messages = [line.split(":", 1)[-1].strip() for line in lines]

summary_lines = []
i = 0
while i < len(messages):
    current = messages[i]
    count = 1
    # count consecutive identical messages
    while i + count < len(messages) and messages[i + count] == current:
        count += 1
    # append with count if >1
    if count > 1:
        summary_lines.append(f"{current}  (x{count})")
    else:
        summary_lines.append(current)
    i += count  # skip counted lines

# write summary to new file
with open(output_file, "w") as f:
    for line in summary_lines:
        f.write(line + "\n")


```

```python
print(f"Grouped consecutive summary saved to {output_file}")
```

```python
from collections import Counter

input_file = "executed_output.txt"
output_file = "grouped_summary.txt"

# read file
with open(input_file, "r") as f:
    lines = [line.strip() for line in f]

# extract only the message part (after colon)
messages = [line.split(":", 1)[-1].strip() for line in lines]

# count occurrences
counts = Counter(messages)

# write summary to new file
with open(output_file, "w") as f:
    for msg, count in counts.items():
        if count > 1:
            f.write(f"{msg}  (x{count})\n")
        else:
            f.write(f"{msg}\n")


```

```python
print(f"Grouped summary saved to {output_file}")
```
